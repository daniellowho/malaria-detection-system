```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("iarunava/cell-images-for-detecting-malaria")

print("Path to dataset files:", path)
```

```python
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. Define the correct subdirectory from your kagglehub path
# The kagglehub download usually contains an extra 'cell_images' nesting level
data_path = os.path.join(path, 'cell_images', 'cell_images')

# 2. Initialize the ImageDataGenerator
# This handles the 'Reading' logic, normalization, and splitting
datagen = ImageDataGenerator(
    rescale=1./255,           # Normalizes pixels from [0-255] to [0-1]
    validation_split=0.2      # Automatically reserves 20% for testing
)

# 3. Read the Training Dataset
train_generator = datagen.flow_from_directory(
    data_path,
    target_size=(224, 224),   # Standard input size for MobileNetV2
    batch_size=32,
    class_mode='binary',      # Parasitized vs Uninfected
    subset='training',
    shuffle=True
)

# 4. Read the Validation Dataset
validation_generator = datagen.flow_from_directory(
    data_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary',
    subset='validation',
    shuffle=False
)

# 5. Output summary of the reading process
print(f"Found {train_generator.samples} images for training.")
print(f"Found {validation_generator.samples} images for validation.")
print(f"Class indices: {train_generator.class_indices}")
```
